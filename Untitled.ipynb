Заполнение колонок middle_avg_bill и middle_coffee_cup, если они не заполнены в соответствующих местах.

```python
def calc_middle(avg_bill: str) -> float:
    """
    Функция для подсчета среднего значения для колонок middle_avg_bill и middle_coffee_cup.

    :param avg_bill: значение колонки avg_bill.
    :return: значение для колонки middle_avg_bill/middle_coffee_cup.
    """
    result = match(r"(\d+) ?– ?(\d+).+", avg_bill)
    if result:
        min_bill = int(result.group(1))
        max_bill = int(result.group(2))
        return (min_bill + max_bill) / 2
    result = search(r"\d+", avg_bill)
    if result:
        return float(result.group(0))
    return np.nan 
```

```python
def fill_middle_bill(row: pd.Series, title: str, filled_column_name: str) -> float:
    """
    Функция для заполнения пустых значений в колонках middle_avg_bill/middle_coffee_cup.

    :param row: обрабатываемая строчка.
    :param title: заголовок, который будет искаться в колонке avg_bill.
    :param filled_column_name: имя заполняемой колонки.
    :return: значение для заполняемой колонки.
    """
    avg_bill: str = str(row["avg_bill"])
    filled_column_value = row[filled_column_name]
    if title in avg_bill and not filled_column_value:
        _, _, value = avg_bill.partition(":")
        middle = calc_middle(value)
        return middle
    return filled_column_value
```

```python
df["middle_avg_bill"] = df.apply(fill_middle_bill, axis=1, args=("Средний счёт", "middle_avg_bill"))
```

```python
df["middle_coffee_cup"] = df.apply(fill_middle_bill, axis=1, args=("Цена чашки капучино", "middle_coffee_cup"))
```

#### Создание столбца street с названиями улиц из столбца с адресом.

```python
def street_from_address(address: str) -> str | None:
    """
    Функция для выделения улицы из адреса.

    :param address: адрес.
    :return: название улицы, если удалось выделить.
    """
    address_parts = address.split(",")
    if len(address_parts) >= 2:
        return address_parts[1].strip()
    return None
```

```python
df["street"] = df["address"].apply(street_from_address)
```

#### Создание столбца is_24_7 с обозначением, что заведение работает ежедневно и круглосуточно (24/7):
  
* логическое значение True — если заведение работает ежедневно и круглосуточно;
* логическое значение False — в противоположном случае.

```python
def is_24_7(v):
    """Функция для заполнения столбца is_24_7"""
    if isinstance(v, str):
        return v.find("ежедневно, круглосуточно") != -1
    else:
        return False
```

```python
df["is_24_7"] = df["hours"].apply(is_24_7)
```

### 4. Какие категории заведений чаще являются сетевыми? Исследуйте данные и ответьте на вопрос графиком.

```python
categories_chain = df.groupby(['category', 'chain']).size().reset_index(name='amount')
categories_chain['chain'] = categories_chain['chain'].replace({0: 'несетевое', 1: 'сетевое'})
categories_chain = categories_chain.sort_values(by='amount', ascending=False)
categories_chain.head()
```

```python
categories_chain.info
```

```python
pivot_table = categories_chain.pivot_table(
    index='category',
    columns='chain',
    values='amount',
    aggfunc='sum'
).fillna(0)
pivot_table.head()
```

```python
# Нормируем по строкам (каждая категория = 100%)
pivot_percentage = pivot_table.div(pivot_table.sum(axis=1), axis=0) * 100

# Добавляем суффиксы к названиям столбцов
pivot_percentage_with_abs = pivot_percentage.copy()
pivot_table_with_suffix = pivot_table.add_suffix('_abs')

# Объединяем
result_with_abs = pd.concat([pivot_percentage, pivot_table_with_suffix], axis=1)
result_with_abs['total'] = result_with_abs['сетевое_abs'] + result_with_abs['несетевое_abs']

# Сортируем
result_with_abs = result_with_abs.sort_values(by='несетевое')
result_with_abs.head()
```

```python
df_for_plot = result_with_abs[['несетевое', 'сетевое']]

# Строим горизонтальный stacked bar chart
fig, ax = plt.subplots(figsize=(12, 8))

# Используем barh вместо bar
bars = df_for_plot.plot(kind='barh', stacked=True, ax=ax)
plt.title('Распределение сетевых/несетевых заведений в каждой категории', fontsize=14)
plt.xlabel('Доля, %')
plt.ylabel('Категория')

ax.legend(loc='lower right', title='Тип заведения')

# Добавляем подписи значений
for i, container in enumerate(ax.containers):
    # Для stacked bar лучше показывать значения только если они достаточно большие
    labels = [f'{val:.1f}%' if val > 5 else '' for val in container.datavalues]
    ax.bar_label(container, labels=labels, label_type='center')

plt.tight_layout()
plt.show()
```

Более высокая доля несетевых заведений наблюдается в категориях, которые часто связаны с локальным, традиционным или специализированным форматом:

Более высокая доля сетевых заведений характерна для категорий, которые легче масштабируются и стандартизируются:

#### 7. Постройте фоновую картограмму (хороплет) со средним рейтингом заведений каждого района. Границы районов Москвы, которые встречаются в датасете, хранятся в файле admin_level_geomap.geojson (скачать файл для локальной работы).

```python
district_rating = df.groupby('district', as_index=False)['rating'].mean().sort_values(by='rating', ascending=False).reset_index(drop=True)
district_rating.head()
```

```python
url = 'https://code.s3.yandex.net/data-analyst/admin_level_geomap.geojson'
# Загружаем GeoJSON файл с границами районов
gdf = gpd.read_file(url)

gdf.head(10)
```

```python
gdf.info()
```

```python
# Объединяем данные
gdf_merged = gdf.merge(district_rating, 
                      left_on='name', 
                      right_on='district',
                      how='left')
gdf_merged.head()
```

```python
# Создаем фигуру
fig, ax = plt.subplots(1, 1, figsize=(15, 12))

# Построение хороплета
gdf_merged.plot(column='rating', 
                ax=ax, 
                legend=True,
                legend_kwds={
                    'label': "Средний рейтинг заведений",
                    'orientation': "vertical",
                    'shrink': 0.7,
                    'aspect': 40
                },
                cmap=ListedColormap(list(reversed(palette))),
                edgecolor='black',
                linewidth=0.5,
                alpha=0.8,
                missing_kwds={
                    "color": "lightgrey",
                    "edgecolor": "black",
                    "hatch": "///",
                    "label": "Нет данных"
                })

# Настройка внешнего вида
ax.set_title('Средний рейтинг заведений по районам Москвы', 
             fontsize=18, pad=20, fontweight='bold')

# Добавляем подписи для районов с данными
for idx, row in gdf_merged.iterrows():
    if not pd.isna(row['rating']):
        # Получаем центр полигона
        centroid = row.geometry.centroid
        # Добавляем подпись с названием и рейтингом
        ax.annotate(text=f"{row['name']}\n{row['rating']:.2f}", 
                   xy=(centroid.x, centroid.y),
                   ha='center',
                   va='center',
                   fontsize=8,
                   bbox=dict(boxstyle="round,pad=0.3", 
                            facecolor='white', 
                            edgecolor='gray', 
                            alpha=0.8))

ax.set_axis_off()
plt.tight_layout()
plt.show()
```

#### 8. Отобразите все заведения датасета на карте с помощью кластеров средствами библиотеки folium.

```python
location = df[['name', 'lat', 'lng']]
location.head()
```

```python
# Создаем базовую карту (центрируем по средним координатам)
center_lat = location['lat'].mean()
center_lng = location['lng'].mean()
m = folium.Map(location=[center_lat, center_lng], zoom_start=12)

# Создаем кластер маркеров
marker_cluster = MarkerCluster().add_to(m)

# Добавляем маркеры для каждого заведения
for idx, row in location.iterrows():
    # Создаем popup с названием заведения
    popup_text = f"<b>{row['name']}</b><br>Широта: {row['lat']:.6f}<br>Долгота: {row['lng']:.6f}"
    popup = folium.Popup(popup_text, max_width=300)
    
    # Добавляем маркер в кластер
    folium.Marker(
        location=[row['lat'], row['lng']],
        popup=popup,
        tooltip=row['name']
    ).add_to(marker_cluster)

# Отображаем карту
m
```
